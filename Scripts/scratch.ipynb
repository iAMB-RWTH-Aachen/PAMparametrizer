```python
import os
if os.path.split(os.getcwd())[-1] != 'PAM_Parametrization': os.chdir('..')

from PAMparametrizer.utils.pam_generation import create_pamodel_from_diagnostics_file, setup_pputida_pam
from pathlib import Path
from cobra.io import read_sbml_model
import pandas as pd
import numpy as np
```

```python
PAM_MULTI_FILE = Path('Results/2_parametrization/proteinAllocationModel_iJN1463_EnzymaticData_multi.xlsx')
c_uptake_id = 'EX_glc__D_e'
dfile = Path('Results/2_parametrization/diagnostics/pam_parametrizer_diagnostics_iJN1463_allbatch_1.xlsx')
pam_file = Path('Results/3_analysis/parameter_files/proteinAllocationModel_EnzymaticData_iJN1463_allexp_1.xlsx')

MODEL_FILE = Path('Models/iJN1463.xml')
```

```python
gem = read_sbml_model(MODEL_FILE)
ref_pam = setup_pputida_pam(PAM_MULTI_FILE,
                            gem.copy(),
                            sensitivity=False)

pam = setup_pputida_pam(pam_file,
                            gem.copy(),
                            sensitivity=False)

dpam = create_pamodel_from_diagnostics_file(file_path=dfile,
                                                   model=ref_pam.copy(copy_with_pickle=True),
                                                   substrate_uptake_id=c_uptake_id,
                                                   enzyme_sector_update=False,
                                                   other_enzyme_id_pattern=ref_pam.configuration.ENZYME_ID_REGEX
                                                   )
```

```python
for model in [gem, ref_pam, dpam]:
    model.reactions.get_by_id(c_uptake_id).lower_bound = -6.14
    model.optimize()
    print(model.summary())
```
